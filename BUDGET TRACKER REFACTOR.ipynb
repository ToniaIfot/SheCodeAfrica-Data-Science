```python
'''Budget tracker: list of my expenses and my budget for each category'''
expenses = [
    {"category": "Food", "item": "Rice", "amount": 9500},
    {"category": "Food", "item": "Beans", "amount": 4000},
    {"category": "Food", "item": "Tomatoes", "amount": 3500},
    {"category": "Transport", "item": "Bus fare", "amount": 2500},
    {"category": "Transport", "item": "Bolt ride", "amount": 4500},
    {"category": "Airtime", "item": "MTN recharge", "amount": 2000},
    {"category": "Airtime", "item": "Data bundle", "amount": 3500},
    {"category": "Rent", "item": "Monthly rent", "amount": 30000},
    {"category": "Food", "item": "Bread", "amount": "n/a"},
    {"category": "Transport", "item": "Okada", "amount": None},
]
'''function to change the amount to a number, it raises an error if the amount is missing'''
def to_number(value):
    if value == None:
        raise ValueError("amount is missing")
    return float(value)

'''function to add up the spending for each category, it skips bad amounts so the code does not crash'''
def total_by_category(expenses):
    totals = {}
    for e in expenses:
        try:
            amount = to_number(e["amount"])
            category = e["category"].title()
            if category in totals:
                totals[category] = totals[category] + amount
            else:
                totals[category] = amount
        except ValueError as err:
            print("skipped", e["item"], "because", err)
    return totals

'''function to print the summary and return the categories that are over budget'''
def summarise(totals, budgets):
    over_budget = []
    for category in budgets:
        if category in totals:
            spent = totals[category]
        else:
            spent = 0
        budget = budgets[category]
        if spent > budget:
            status = f"OVER by {spent - budget:,.2f}"
            over_budget.append(category)
        else:
            status = "OK"
        print(f"{category}: spent NGN {spent:,.2f} of NGN {budget:,.2f} {status}")
    return over_budget

'''function using a list comprehension to get the expenses above a threshold'''
def expenses_above(expenses, threshold):
    good_expenses = [e for e in expenses if e["amount"] != None and e["amount"] != "n/a"]
    return [e for e in good_expenses if e["amount"] > threshold]
budgets = {"Food": 15000, "Transport": 8000, "Airtime": 5000, "Rent": 30000}
```

```python
'''calling the functions to get my totals, summary and the big expenses'''
totals = total_by_category(expenses)

print("------ BUDGET SUMMARY ------")
over = summarise(totals, budgets)
print("Over budget:", over)

print("------ EXPENSES ABOVE NGN 5,000 ------")
big_expenses = expenses_above(expenses, 5000)
for e in big_expenses:
    print(f"{e['item']}: NGN {e['amount']:,.2f}")
```
