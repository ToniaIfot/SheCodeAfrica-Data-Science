```python
item1_name = input("What did you buy? ")
item1_amount = float(input("How much does that cost? "))
item1_quantity = int(input("How many did you get? "))
item1_total = item1_amount * item1_quantity
print("okay, the total for that is", item1_total,"naira")

item2_name = input("What did you buy? ")
item2_amount = float(input("How much does that cost? "))
item2_quantity = int(input("How many did you get? "))
item2_total = item2_amount * item2_quantity
print("okay, the total for that is", item2_total,"naira")

item3_name = input("What did you buy? ")
item3_amount = float(input("How much does that cost? "))
item3_quantity = int(input("How many did you get? "))
item3_total = item3_amount * item3_quantity
print("okay, the total for that is", item3_total,"naira") 
```

```python
subtotal = item1_total + item2_total + item3_total

if subtotal >= 20000:
    discount = subtotal * 0.10
elif subtotal >= 10000:
    discount = subtotal * 0.05
else:
    discount = 0

final_total = subtotal - discount

print("------ RECEIPT ------")
print(f"{item1_name} x{item1_quantity}: NGN {item1_total:,.2f}")
print(f"{item2_name} x{item2_quantity}: NGN {item2_total:,.2f}")
print(f"{item3_name} x{item3_quantity}: NGN {item3_total:,.2f}")
print(f"Subtotal: NGN {subtotal:,.2f}")
print(f"Discount: NGN {discount:,.2f}")
print(f"TOTAL: NGN {final_total:,.2f}")
```
