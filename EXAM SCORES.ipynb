```python
'''Importing thee numpy library'''
import numpy as np
```

```python
'''names of student'''
Student_names = np.array(["Jude", "Pelumi", "Sommie", "Miley", "Vicky"])

'''Subject names'''
Subject = np.array(["Petroleum geology", "Applied geophysics", "Engineering geology", "Petrophysics"])

'''Exam scores with the students being on the row and subjects on the column'''
Scores = np.array ([
    [70, 65, 82, 90],
    [67, 80, 91, 45],
    [89, 67, 65, 60],
    [28, 45, 71, 43],
    [70, 71, 56, 87]])
print(Scores.shape)
```

```python
'''Average score of the students'''
'''Average score of each student (axis=1 averages across each row, and each row is a student)'''
mean_score = Scores.mean(axis=1)
print(Student_names)
print(mean_score)
```

```python
'''Min and Max score for each subject (axis=0 works down each column, and each column is a subject)'''
Min_Score = Scores.min(axis=0)
Max_Score = Scores.max(axis=0)
print(Min_Score)
print(Max_Score)

'''Min-max normalisation: scales each subject to between 0 and 1, without a loop (broadcasting)'''
normalised = (Scores - Min_Score) / (Max_Score - Min_Score)
print(np.round(normalised, 2))
```

```python
'''Class average (average of all the scores)'''
class_average = Scores.mean()
print(f"Class average: {class_average:.2f}")

'''Creating a mask: True for students whose average is above the class average'''
mask = mean_score > class_average
print(mask)

'''Using the mask on the names to list the students above the class average'''
print("Students above the class average:", Student_names[mask])
```
