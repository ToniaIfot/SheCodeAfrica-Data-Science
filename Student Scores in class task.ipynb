```python
'''student scores in-class task'''

students = [
    {"name": "Tolu", "scores": [80, 75, 88]},
    {"name": "Emeka", "scores": [55, 60, 52]},
    {"name": "Zainab", "scores": [92, 89, 95]},
    {"name": "Kemi", "scores": [68, 72, 65]},
    {"name": "Uche", "scores": [74, 70, 79]},
    {"name": "Halima", "scores": [85, 90, 82]},
]
```

```python
'''average of each student'''

averages = {}
for s in students:
    avg = sum(s["scores"]) / len(s["scores"])
    averages[s["name"]] = round(avg, 1)

for name in averages:
    print(name, "average is", averages[name])
```

```python
'''top three students'''

pairs = []
for name in averages:
    pairs.append((averages[name], name))

pairs.sort(reverse=True)

print("top three students:")
for avg, name in pairs[:3]:
    print(name, avg)
```

```python
'''counting the grade bands'''

bands = {"A": 0, "B": 0, "C": 0, "F": 0}
for avg in averages.values():
    if avg >= 80:
        bands["A"] += 1
    elif avg >= 70:
        bands["B"] += 1
    elif avg >= 60:
        bands["C"] += 1
    else:
        bands["F"] += 1

print(bands)
```
