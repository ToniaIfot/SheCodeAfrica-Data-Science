```python
'''my expenses for the month, each one is a dictionary'''
expenses = [
    {"category": "Skincare", "item": "Sunscreen", "amount": 30000},
    {"category": "Clothes", "item": "Corporate wears", "amount": 60000},
    {"category": "Airtime", "item": "Airtel monthly plan", "amount": 7500},
    {"category": "Food", "item": "Provisions", "amount": 120000},
    {"category": "Outings", "item": "Hangout with friends", "amount": 50000},
    {"category": "Self care", "item": "Nails and lashes", "amount": 30000},
    {"category": "Family", "item": "Mom", "amount": 50000},
    {"category": "Investment", "item": "Nigerian stock", "amount": 30000},
]

'''my budget for each category'''
budgets = {
    "Skincare": 50000,
    "Clothes": 30000,
    "Airtime": 7500,
    "Food": 100000,
    "Outings": 50000,
    "Self care": 20000,
    "Family": 100000,
    "Novels": 15000,
    "Investment": 30000,
    "Savings": 30000,
}
```

```python
'''adding up how much I spent in each category'''
totals = {}
for e in expenses:
    category = e["category"]
    amount = e["amount"]
    '''if the category already has a total, add to it, if not, start a new total'''
    if category in totals:
        totals[category] = totals[category] + amount
    else:
        totals[category] = amount

print(totals)
```

```python
'''comparing what I spent in each category with my budget and printing a summary'''
print("------ MY BUDGET SUMMARY ------")
for category in budgets:
    '''categories I did not spend on (like Novels and Savings) are not in totals, so they get 0'''
    if category in totals:
        spent = totals[category]
    else:
        spent = 0
    budget = budgets[category]
    '''flagging the categories where I spent more than my budget'''
    if spent > budget:
        status = f"OVER by {spent - budget:,}"
    else:
        status = "OK"
    print(f"{category} spent {spent:,} of {budget:,} {status}")

'''overall total for the month'''
total_spent = sum(totals.values())
total_budget = sum(budgets.values())
print(f"Total spent {total_spent:,} of {total_budget:,}")
```
